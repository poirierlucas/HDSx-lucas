# BMI Summary Parity: Python/pandas

Run this notebook from the repository root. It loads the same NHANES CSV as `polyglot-parity.qmd`, excludes rows missing BMI, `IncomeGroup`, or `Gender`, and reports the same count, mean, median, and sample standard deviation by group. The reusable summary function is imported from `helpers.py`.

```python
from pathlib import Path
import sys

import pandas as pd

submission_dir = Path("assignments/assignment04-polyglot/submission").resolve()
sys.path.insert(0, str(submission_dir))
from helpers import summarize_bmi_by_income_gender

data_path = Path("examples/nhanes-equity/data/nhanes_equity_v6.csv")
nhanes = pd.read_csv(data_path, low_memory=False)
bmi_summary = summarize_bmi_by_income_gender(nhanes)

print(f"Raw rows: {len(nhanes):,}")
print(f"Raw columns: {len(nhanes.columns):,}")
print(f"Complete cases: {bmi_summary['n'].sum():,}")
print(f"Groups: {len(bmi_summary):,}")
bmi_summary.round({"mean_bmi": 2, "median_bmi": 2, "sd_bmi": 2})
```
